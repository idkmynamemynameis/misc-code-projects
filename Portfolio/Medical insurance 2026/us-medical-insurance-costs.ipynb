```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.read_csv('insurance.csv')
df.head()
#sewt up lists for each region and smoking status
nwns=[]
north_east_non_smoker=[]
swns=[]
sens=[]
nwws=[]
news=[]
swws=[]
sews=[]
#added rows to lists based on region and smoking status
for row in df.itertuples():
    if row[6]=='northwest' and row[5]=='no':
        nwns.append(row)
    elif row[6]=='northeast'and row[5]=='no':
        north_east_non_smoker.append(row)
    elif row[6]=='southwest'and row[5]=='no':
        swns.append(row)
    elif row[6]=='southeast'and row[5]=='no':
        sens.append(row)
for row in df.itertuples():
    if row[6]=='northwest' and row[5]=='yes':
        nwws.append(row)
    elif row[6]=='northeast'and row[5]=='yes':
        news.append(row)
    elif row[6]=='southwest'and row[5]=='yes':
        swws.append(row)
    elif row[6]=='southeast'and row[5]=='yes':
        sews.append(row)

def get_average_cost(list):
    total_cost=0
    for row in list:
        total_cost+=row[7]
    return total_cost/len(list)
nwns_avg_cost=get_average_cost(nwns)
north_east_non_smoker_avg_cost=get_average_cost(north_east_non_smoker)
swns_avg_cost=get_average_cost(swns)
sens_avg_cost=get_average_cost(sens)
nwws_avg_cost=get_average_cost(nwws)
news_avg_cost=get_average_cost(news)
swws_avg_cost=get_average_cost(swws)
sews_avg_cost=get_average_cost(sews)

main_graph='''
labels=['Northwest No','Northeast No','Southwest No','Southeast No','Northwest Yes','Northeast Yes','Southwest Yes','Southeast Yes']
avg_costs=[nwns_avg_cost,north_east_non_smoker_avg_cost,swns_avg_cost,sens_avg_cost,nwws_avg_cost,news_avg_cost,swws_avg_cost,sews_avg_cost]
plt.bar(labels,avg_costs)
plt.xlabel('Region and Smoking Status')
plt.ylabel('Average Insurance Cost')
plt.title('Average Insurance Cost by Region and Smoking Status')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
'''
difference_graph='''
labels=['Northwest','Northeast','Southwest','Southeast']

avg_costs_difference=[nwws_avg_cost-nwns_avg_cost,news_avg_cost-north_east_non_smoker_avg_cost,swws_avg_cost-swns_avg_cost,sews_avg_cost-sens_avg_cost]
x = range(len(labels))
plt.bar(x, avg_costs_difference, width=0.4, label='Difference', align='center')
plt.xlabel('Region')
plt.ylabel('Average Insurance Cost Difference (Smokers - Non-Smokers)')
plt.title('Average Insurance Cost by Region and Smoking Status')
plt.xticks(x, labels)
plt.legend()
plt.tight_layout()
plt.show()
'''
#exec(main_graph)
```

```python
nwm=[]
nwf=[]
nem=[]
nef=[]
swm=[]
swf=[]
sem=[]
sef=[]
for row in df.itertuples():
    if row[6]=='northwest' and row[2]=='male':
        nwm.append(row)
    elif row[6]=='northwest' and row[2]=='female':
        nwf.append(row)
    elif row[6]=='northeast' and row[2]=='male':
        nem.append(row)
    elif row[6]=='northeast' and row[2]=='female':
        nef.append(row)
    elif row[6]=='southwest' and row[2]=='male':
        swm.append(row)
    elif row[6]=='southwest' and row[2]=='female':
        swf.append(row)
    elif row[6]=='southeast' and row[2]=='male':
        sem.append(row)
    elif row[6]=='southeast' and row[2]=='female':
        sef.append(row)

nwm_avg_cost=get_average_cost(nwm)
nwf_avg_cost=get_average_cost(nwf)
nem_avg_cost=get_average_cost(nem)
nef_avg_cost=get_average_cost(nef)
swm_avg_cost=get_average_cost(swm)
swf_avg_cost=get_average_cost(swf)
sem_avg_cost=get_average_cost(sem)
sef_avg_cost=get_average_cost(sef)

# Region and gender chart
plt.figure(figsize=(10,5))
labels_gender = ['Northwest Male','Northwest Female','Northeast Male','Northeast Female','Southwest Male','Southwest Female','Southeast Male','Southeast Female']
avg_costs_gender = [nwm_avg_cost,nwf_avg_cost,nem_avg_cost,nef_avg_cost,swm_avg_cost,swf_avg_cost,sem_avg_cost,sef_avg_cost]
plt.bar(labels_gender, avg_costs_gender)
plt.xlabel('Region and Gender')
plt.ylabel('Average Insurance Cost')
plt.title('Average Insurance Cost by Region and Gender')
plt.xticks(rotation=45)
plt.tight_layout()
#plt.show()

# Grouped bar chart by region with identical category order in each area
avg_by_region_gender_smoker = df.groupby(['region','sex','smoker'])['charges'].mean()
region_order = ['northwest','northeast','southwest','southeast']
category_order = [('male','no'), ('female','no'), ('male','yes'), ('female','yes')]
category_labels = ['Male Non-Smoker','Female Non-Smoker','Male Smoker','Female Smoker']
x = np.arange(len(region_order))
width = 0.18
plt.figure(figsize=(12,6))
for i, cat in enumerate(category_order):
    values = [avg_by_region_gender_smoker.loc[(region, cat[0], cat[1])] for region in region_order]
    plt.bar(x + i*width, values, width, label=category_labels[i])
plt.xlabel('Region')
plt.ylabel('Average Insurance Cost')
plt.title('Average Cost by Region and Gender/Smoking Group')
plt.xticks(x + 1.5*width, ['Northwest','Northeast','Southwest','Southeast'])
plt.legend()
plt.tight_layout()
#plt.show()

# Overall male/female smoker vs non-smoker chart
avg_by_sex_smoker = df.groupby(['sex','smoker'])['charges'].mean()
labels_small = ['Male Non-Smoker', 'Female Non-Smoker', 'Male Smoker', 'Female Smoker']
avg_costs_small = [
    avg_by_sex_smoker.loc[('male','no')],
    avg_by_sex_smoker.loc[('female','no')],
    avg_by_sex_smoker.loc[('male','yes')],
    avg_by_sex_smoker.loc[('female','yes')]
]
plt.figure(figsize=(8,5))
plt.bar(labels_small, avg_costs_small, color=['#1f77b4','#ff7f0e','#2ca02c','#d62728'])
plt.xlabel('Group')
plt.ylabel('Average Insurance Cost')
plt.title('Average Cost: Male/Female, Smoker vs Non-Smoker')
plt.tight_layout()
#plt.show()
```
